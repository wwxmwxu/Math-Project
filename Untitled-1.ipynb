```python
def calculate_math_demand_metrics(digits1: int, digits2: int, num_carries: int) -> dict:

    max_digits = max(digits1, digits2)
    stage_number = max_digits
    
    # Track interim values for each stage
    during_values = []
    after_values = []
    
    carries_remaining = num_carries
    
    for stage in range(stage_number):
        # DURING this stage calculation
        values_during = 0
        
        # If there's a carry from previous stage, we need to remember it
        if stage > 0 and carries_remaining > 0:
            values_during += 1  # carry value
            carries_remaining -= 1
        
        during_values.append(values_during)
        
        # AFTER this stage calculation
        if stage < stage_number - 1:  # Not the final stage
            values_after = 0
            
            # Need to remember the result digit(s) we just calculated
            values_after += 1  # unit place result
            values_after += 1  # place marker
            
            # If this stage produces a carry for next stage, remember it
            if carries_remaining > 0:
                values_after += 1  # tens carry value
                values_after += 1  # carry marker
            
            after_values.append(values_after)
        else:
            # Final stage - just the complete result
            values_after = max_digits  # all digit positions of final answer
            after_values.append(values_after)
    
    # Calculate totals
    total_during = sum(during_values)
    total_after = sum(after_values)
    
    # Calculate averages
    avg_during = total_during / stage_number
    avg_after = total_after / stage_number
    
    # Calculate sums
    sum_avg = avg_during + avg_after
    sum_total = total_during + total_after
    
    return {
        'avg_interim_during': avg_during,
        'total_during': total_during,
        'avg_interim_after': avg_after,
        'total_after': total_after,
        'sum_avg': sum_avg,
        'sum_total': sum_total
    }


def get_metrics(digits1: int, digits2: int, num_carries: int) -> tuple:
    """
    Returns: (avg_during, total_during, avg_after, total_after, sum_avg, sum_total)
    """
    result = calculate_math_demand_metrics(digits1, digits2, num_carries)
    return (
        result['avg_interim_during'],
        result['total_during'],
        result['avg_interim_after'],
        result['total_after'],
        result['sum_avg'],
        result['sum_total']
    )


# Test with the examples from your description
if __name__ == "__main__":
    print("Example 1: 12+9 (written)")
    print("Expected from Table 1:")
    print("  Total during: 2, Total after: 6")
    print("  Avg during: 1, Avg after: 3, Sum Avg: 4\n")
    
    metrics = calculate_math_demand_metrics(2, 1, 1)
    print("Calculated:")
    for key, value in metrics.items():
        print(f"  {key}: {value}")
    
    print("\n" + "="*70)
    print("Testing with various inputs:\n")
    
    test_cases = [
        (2, 1, 0, "12+6 or 16-3"),
        (2, 1, 1, "12+9 or 23-8"),
        (3, 2, 1, "563+72"),
        (2, 2, 1, "65+16"),
        (2, 2, 2, "98+16"),
        (3, 2, 2, "563+78"),
        (3, 3, 3, "236+776"),
    ]
    
    print(f"{'Digits':<10} {'Carries':<8} {'Avg During':<12} {'Total During':<13} {'Avg After':<11} {'Total After':<12} {'Sum Avg':<9} {'Sum Total':<10} {'Example'}")
    print("-" * 120)
    
    for d1, d2, carries, example in test_cases:
        metrics = calculate_math_demand_metrics(d1, d2, carries)
        digits_str = f"{d1}&{d2}"
        print(f"{digits_str:<10} {carries:<8} "
              f"{metrics['avg_interim_during']:<12.2f} {metrics['total_during']:<13.0f} "
              f"{metrics['avg_interim_after']:<11.2f} {metrics['total_after']:<12.0f} "
              f"{metrics['sum_avg']:<9.2f} {metrics['sum_total']:<10.0f} {example}")
```
